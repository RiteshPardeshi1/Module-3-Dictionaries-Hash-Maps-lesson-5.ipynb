```python
for employee_id, details in employees.items():

    if details["salary"] > 90000:
        print(details["name"], details["salary"])
```

```python
total_salary = 0

for x in employees.values():
    total_salary += x['salary']

print(total_salary)
```

```python
high_paid = {}

for x,y in employees.items():
    if y['salary'] > 90000:
        high_paid[x] = y

print(high_paid)
```

```python
highest_employee = None
highest_salary = float("-inf")

for x,y in employees.items():
    if y['salary'] > highest_salary:
        highest_salary = y['salary']
        highest_employee = y['name']
print(highest_employee,highest_salary)
```
